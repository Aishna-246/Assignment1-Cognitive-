```python

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """My fav topic would be art . There are mulptiple art forms art is a medium to express yourself and your feelings
. Anything that makes you feel something is classified as art . Be it cooking , writing , acting
drawing , painting , singing , dancing etc. As a race we are not just focused on survival but
contemplating , experiencing and making connections with others is a part of being human . """

print("Original Text:\n", text)


```

```python
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("\nCleaned Text:\n", clean_text)
```

```python
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:\n", words)
print("\nSentence Tokens:\n", sentences)
```

```python
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nUsing split():\n", split_words)
print("\nUsing word_tokenize():\n", nltk_words)

print("\nAre both results equal?", split_words == nltk_words)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in nltk_words
    if word.isalpha() and word not in stop_words
]

print("\nWords after removing stopwords:\n", filtered_words)

```

```python
frequency = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")
```

```python
print("\nFrequency Dictionary:\n", dict(frequency))
```

```python
#Question 2


import re
import nltk
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """My fav topic would be art. There are multiple art forms.
Art is a medium to express yourself and your feelings.
Anything that makes you feel something is classified as art.
Be it cooking, writing, acting, drawing, painting, singing, dancing, etc.
As a race we are not just focused on survival, but contemplating,
experiencing and making connections with others is a part of being human."""


words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("Alphabet-only words:\n", words)

```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]
print("\nAfter Stopword Removal:\n", filtered_words)
```

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\nStemmed Words:\n", stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]
print("\nLemmatized Words:\n", lemmatized_words)
```

```python
print("\nComparison:")
for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:15} -> Stem: {stem:12} Lemma: {lemma}")
```

```python
#Question 3
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

# 3 short texts of our own
texts = [
    "Yemen Plunges Back into “Full-Scale Civil War,” Houthis Trade Strikes with Saudi Arabia",
    "“Musk”: Alex Gibney on New Film, Legal Threats & Elon Musk Spending $366M to Help GOP in Midterms",
    "Protesters Say India’s Democracy Is “Rotten” as Modi Gov’t Purges Voter Rolls, Arrests Thousands"
]

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("Bag of Words Representation:")
print(bow_df)
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("\nTF-IDF Scores:")
print(tfidf_df.round(3))
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTop 3 Keywords from Each Text:")
for i, text in enumerate(texts):
    scores = tfidf_matrix[i].toarray().flatten()
    top_indices = scores.argsort()[::-1][:3]

    keywords = [
        (feature_names[j], round(scores[j], 3))
        for j in top_indices
    ]

    print(f"\nText {i + 1}: {text}")
    print("Top keywords:", keywords)
```

```python
#Question 4

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """
Artificial intelligence enables machines to learn from data.
AI systems can recognize patterns and make predictions.
Machine learning is an important part of artificial intelligence.
AI is used in healthcare, education, and automation.
"""

text2 = """
Blockchain stores information in a distributed digital ledger.
Blockchain technology provides transparency and security.
Cryptocurrency transactions often use blockchain networks.
Blockchain records are difficult to modify once verified.
"""

def preprocess(text):
    text = text.lower()
    words = re.findall(r'\b[a-z]+\b', text)
    return words

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI Tokens:\n", tokens1)
print("\nBlockchain Tokens:\n", tokens2)
```

```python
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))
```

```python
documents = [text1, text2]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

cosine_sim = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine_sim, 4))
```

```python
print("\nAnalysis:")
if jaccard > cosine_sim:
    print("Jaccard similarity is higher than cosine similarity.")
elif cosine_sim > jaccard:
    print("Cosine similarity is higher than Jaccard similarity.")
else:
    print("Both similarity measures give the same score.")
```

```python
#Question 5

from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "These headphones have excellent sound quality and comfortable ear cushions. I love the battery life.",
    "The sound is terrible and the battery drains quickly. I am very disappointed.",
    "The headphones look good, but the sound quality is average.",
    "Amazing headphones with clear audio, great bass, and a long-lasting battery.",
    "The product is overpriced and uncomfortable. The connection keeps dropping."
]

positive_reviews = []

for i, review in enumerate(reviews, start=1):
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(f"\nReview {i}: {review}")
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)

    if sentiment == "Positive":
        positive_reviews.append(review)


if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white",
        colormap="viridis"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews available for the word cloud.")
```

```python

import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Training paragraph (~100 words)
text = """
Technology has changed the way people learn, communicate, and work.
Artificial intelligence helps computers solve problems and recognize patterns.
Students use online platforms to access educational resources.
Machine learning allows systems to learn from data and improve predictions.
Technology also helps doctors diagnose diseases and provide better healthcare.
However, excessive use of technology can affect human relationships.
People should use digital tools responsibly and protect their privacy.
The internet connects people from different parts of the world.
New inventions create opportunities for education, science, and business.
In the future, technology may make everyday tasks easier.
Human creativity and critical thinking will remain important.
Technology should be used to improve lives and solve real problems.
"""

# 2. Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1
token_list = tokenizer.texts_to_sequences([text])[0]

print("Total unique words:", total_words - 1)
print("Tokenized sequence:\n", token_list)

# 3. Create input sequences
input_sequences = []

for i in range(1, len(token_list)):
    n_gram_sequence = token_list[:i + 1]
    input_sequences.append(n_gram_sequence)

# Pad sequences to equal length
max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

# Separate input words (X) and target word (y)
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("\nInput shape:", X.shape)
print("Target shape:", y.shape)

# 4. Build a simple LSTM model
model = Sequential([
    Embedding(total_words, 32, input_length=max_sequence_len - 1),
    LSTM(64),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()

# 5. Train the model
model.fit(X, y, epochs=150, verbose=0)
print("\nModel training completed!")

# 6. Generate text from a seed word
def generate_text(seed_text, next_words=10):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted_probs = model.predict(token_list, verbose=0)[0]

        # Sample the next word using predicted probabilities
        predicted_index = np.random.choice(
            len(predicted_probs),
            p=predicted_probs / predicted_probs.sum()
        )

        output_word = None

        for word, index in tokenizer.word_index.items():
            if index == predicted_index:
                output_word = word
                break

        if output_word is None:
            break

        seed_text += " " + output_word

    return seed_text

# Generate 3 new lines
print("\nGenerated Text:\n")

seed = "technology"

for i in range(3):
    print(generate_text(seed, next_words=10))
```
