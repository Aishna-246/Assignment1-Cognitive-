```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print("Corrected temperatures:", corrected)

fahrenheit = (9/5) * corrected + 32
print("Temperatures in Fahrenheit:", fahrenheit)

readings = corrected[corrected > 32]
print("Readings greater than 32°C:", readings)

count = np.sum(corrected > 32)
print("Count:", count)
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))

print("Mean steps:", np.mean(steps))

print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))

print("Total per day:", np.sum(steps, axis=0))

print("Total per user:", np.sum(steps, axis=1))

position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position (row, column):", position)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

subset = original[1:4]
print("Subset:", subset)

subset[0] = 999
print("Original after modifying subset:", original)
print("Subset:", subset)

copied = original[1:4].copy()
copied[0] = 500
print("Original after modifying copy:", original)
print("Copied array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("3x4 matrix:\n", matrix)

print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("Flatten:", flat_array)
print("Ravel:", ravel_array)

ravel_array[0] = 100
print("Matrix after modifying ravel:\n", matrix)

flat_array[1] = 200
print("Matrix after modifying flatten:\n", matrix)
print("Flatten array:", flat_array)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

XT = X.T
print("Transpose of X:\n", XT)

product = XT @ X
print("X.T @ X:\n", product)

inverse = np.linalg.inv(product)
print("Inverse of X.T @ X:\n", inverse)

beta = inverse @ XT @ y
print("OLS coefficients:", beta)

print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])


new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
