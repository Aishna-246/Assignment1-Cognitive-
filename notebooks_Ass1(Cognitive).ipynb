```python
import random 
random.seed(1024170247)
list = []
for _ in range(100):
    list.append(random.randint(100 , 900))
print(list)

odd = []
oddcount = 0 
even =[]
evencount = 0 

for i in list:
    if (i%2 == 0):
        evencount += 1 
        even.append(i)
    else:
        oddcount += 1 
        odd.append(i)
print(even)
print(evencount)
print(odd)
print(oddcount)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2 , int(n ** 0.5) + 1):
        if(n%i == 0):
            return False 
    return True 

prime = [i for i in list if is_prime(i)] 
print(prime)

dict={}

for num in list:
    if num in dict:
        dict[num]+=1 
    else:
        dict[num]=1

max_freq = 0 
freq = 0 

for n in dict:
    if dict[n] > freq :
        freq = dict[n]
        max_freq = n 
print(max_freq)
print(freq)
```
